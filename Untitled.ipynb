```python
import pandas as pd
import matplotlib.pyplot as plt
```

1. Problem Statement

Topic: Video game sales

Link to data source: https://www.kaggle.com/datasets/gregorut/videogamesales

Main question: What patterns exist in video game sales across genres, platforms, and time?

Sub-questions:
1. How have total global video game sales changed over the years?
2. Which genre has the highest average global sales?
3. Is there a relationship between North American sales and European sales?

```python
df = pd.read_csv('vgsales.csv')
df.shape
```

```python
df.head()
```

2. Data Cleaning

a. Drop unwanted features

'Rank' just repeats the row order based on Global_Sales, so it's not needed.

```python
df = df.drop(columns=['Rank'])
df.shape
```

b. Check and handle missing values

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Year'])
df['Publisher'] = df['Publisher'].fillna('Unknown')
df.isnull().sum()
```

c. Check and handle duplicates

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.shape
```

d. Check data types and correct them

```python
df.dtypes
```

```python
df['Year'] = df['Year'].astype(int)
df.dtypes
```

e. Check categorical variables for inconsistent inputsprint(sorted(df['Genre'].unique()))
print(sorted(df['Platform'].unique()))

```python
print(sorted(df['Genre'].unique()))
print(sorted(df['Platform'].unique()))
```

No inconsistencies found. Genre/platform values are already clean

f. Check numeric variables for outliers

```python
df['Global_Sales'].describe()
```

```python
df.sort_values('Global_Sales', ascending=False).head(5)[['Name', 'Global_Sales']]
```

The highest values are real best-selling games (like Wii Sports), not errors so nothing is removed

3. Exploratory Data Analysis

Question A: How have total global video game sales changed over the years?

```python
yearly_sales = df.groupby('Year')['Global_Sales'].sum()
yearly_sales
```

```python
plt.figure(figsize=(10,5))
plt.plot(yearly_sales.index, yearly_sales.values)
plt.title('Total Global Video Game Sales by Year')
plt.xlabel('Year')
plt.ylabel('Global Sales (millions)')
plt.show()
```

Sales rose over time and peaked around 2008-2009, then declined/fell

Question B: Which genre has the highest average global sales?

```python
avg_by_genre = df.pivot_table(values='Global_Sales', index='Genre', aggfunc='mean').sort_values('Global_Sales', ascending=False)
avg_by_genre
```

```python
plt.figure(figsize=(10,5))
plt.bar(avg_by_genre.index, avg_by_genre['Global_Sales'])
plt.title('Average Global Sales by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Global Sales (millions)')
plt.xticks(rotation=45)
plt.show()
```

Platform games have the highest average global sales of any genre

Question C: Is there a relationship between North American sales and European sales?

```python
df[['NA_Sales', 'EU_Sales']].corr()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['NA_Sales'], df['EU_Sales'], alpha=0.3)
plt.title('NA Sales vs EU Sales')
plt.xlabel('NA Sales (millions)')
plt.ylabel('EU Sales (millions)')
plt.show()
```

There is a "positive relationship". Games that sell well in North America tend to also sell well in Europe
